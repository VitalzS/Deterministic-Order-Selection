```python
import sys
import re # Import re for regular expressions

def select(L, k):
    if len(L) <= 5:
        return sorted(L)[k - 1]

    groups = [L[i:i + 5] for i in range(0, len(L), 5)]
    medians = [sorted(group)[len(group) // 2] for group in groups]

    median_k = (len(medians) + 1) // 2
    M = select(medians, median_k)

    L1 = [x for x in L if x < M]
    L2 = [x for x in L if x == M]
    L3 = [x for x in L if x > M]

    count_L1 = len(L1)
    count_L2 = len(L2)

    if k <= count_L1:
        return select(L1, k)
    elif k <= count_L1 + count_L2:
        return M
    else:
        return select(L3, k - count_L1 - count_L2)

def read_matrix_from_file(filename):
    elements = []
    with open(filename, 'r') as f:
        for line in f:
            cleaned = line.replace('[', ' ').replace(']', ' ').replace('(', ' ').replace(')', ' ').replace(',', ' ')
            tokens = cleaned.strip().split()
            for token in tokens:
                try:
                    val = float(token) if '.' in token else int(token)
                    elements.append(val)
                except ValueError:
                    pass
    return elements

def parse_problem_input(input_str):
    # Example format: {9, {-384510907, 977350089, ...}}
    match = re.match(r"\{(\d+),\s*\{(.*?)\}\}", input_str)
    if not match:
        raise ValueError("Invalid input string format")

    k_val = int(match.group(1))
    list_str = match.group(2)

    elements = []
    for s in list_str.split(','):
        s = s.strip()
        if s:
            elements.append(int(s))
    return k_val, elements

if __name__ == "__main__":
    # Default values
    file_path = "1.txt"
    k = 1

    # Filter sys.argv to remove kernel-specific arguments for interactive environments
    args_to_parse = sys.argv[1:] # Skip the script name itself
    user_args = []

    # Identify and remove ipykernel-specific arguments (like '-f kernel-xxxx.json')
    i = 0
    while i < len(args_to_parse):
        if args_to_parse[i] == '-f':
            i += 2 # Skip '-f' and the following kernel file path
        else:
            user_args.append(args_to_parse[i])
            i += 1

    # Use user-provided arguments if available
    if len(user_args) > 0:
        file_path = user_args[0]
    if len(user_args) > 1:
        try:
            k = int(user_args[1])
        except ValueError:
            print(f"Warning: Could not convert '{user_args[1]}' to an integer for k. Using default k={k}")

    L = read_matrix_from_file(file_path)
    result = select(L, k)
    print(result)

    # The previous demonstration with `example_input_string` is removed to focus on file input.
```

```python
current_file_path = "1.txt"
try:
    L_from_file = read_matrix_from_file(current_file_path)
    sorted_L = sorted(L_from_file)
    print(f"Numbers read from '{current_file_path}' (sorted):\n{sorted_L}")
    if len(sorted_L) >= 1:
        print(f"The 1st smallest element should be: {sorted_L[0]}")
    else:
        print("The file is empty or contains no valid numbers.")
except FileNotFoundError:
    print(f"Error: File '{current_file_path}' not found.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
def solve_kth_smallest(filename, target_k):
    # Read all numbers from the file as the dataset
    dataset = read_matrix_from_file(filename)
    if not dataset:
        print("Error: No elements found in file.")
        return

    print(f"Target k: {target_k}")
    print(f"Dataset size: {len(dataset)}")

    if target_k < 1 or target_k > len(dataset):
        print(f"Error: k={target_k} is out of bounds for dataset of size {len(dataset)}")
        return

    ans = select(dataset, target_k)
    print(f"The {target_k}-th smallest element is: {ans}")

# Run the refactored code (for example, finding the 1st smallest element)
solve_kth_smallest('1.txt', target_k=1)
```

```python
def solve_structured_input(input_str):
    # Extract k and the list of numbers using the parse_problem_input function
    k_val, dataset = parse_problem_input(input_str)

    print(f"Parsed k: {k_val}")
    print(f"Dataset size: {len(dataset)}")

    # Calculate the k-th smallest element
    ans = select(dataset, k_val)
    print(f"Result: {ans}")
    return ans

# Test with the provided example
example_input = "{9, {-384510907, 977350089, -447882330, 1565902303, 1911035140, -2006928812, -1874555214, 1274434577, 1917928800, 768922581}}"
solve_structured_input(example_input)
```
